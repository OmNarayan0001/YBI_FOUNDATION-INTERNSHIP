# Pandas Data Summary

# What is Data Summary?
Data summary refers to the process of extracting meaningful insights from a dataset by calculating statistical measures and aggregating data points. It involves generating descriptive statistics that provide a snapshot of the data's central tendency, dispersion, and distribution. These summaries help in understanding the data's structure, identifying patterns, and making informed decisions. Data summary is grounded in statistical theory, which provides the mathematical foundation for summarizing data. Key statistical measures include:

* **Mean:** The average value of a dataset, representing its central tendency.   

* **Median:** The middle value when data points are sorted, offering a robust measure of central tendency.   

* **Mode:** The most frequently occurring value(s) in a dataset.   

* **Standard Deviation:** A measure of data dispersion around the mean.   

* **Variance:** The square of the standard deviation, indicating data spread.

# Properties of Data Summary
Data summary has several properties that make it a valuable tool in data analysis:   

* **Simplicity:** Provides a clear and concise overview of data characteristics.   

* **Insightful:** Reveals patterns, trends, and anomalies within the data.   

* **Comparative:** Facilitates comparison between different datasets or data groups.   

* **Informative:** Aids in decision-making by highlighting key data points.

# Creating Data Summaries with Pandas   
Pandas is a versatile library that simplifies the process of creating data summaries. It offers various functions to compute descriptive statistics and aggregate data efficiently. Here's a step-by-step guide to creating data summaries using Pandas:

**Step 1: Importing Pandas**

First, import the Pandas library to access its data manipulation capabilities.

```python
import pandas as pd
```

**Step 2: Loading Data **
  
Load your dataset into a Pandas DataFrame. For this example, we'll use a hypothetical dataset of Indian students' exam scores:

```python
data = {
    'Name': ['Aarav', 'Isha', 'Rahul', 'Sneha', 'Ananya'],
    'Math': [85, 92, 78, 88, 95],
    'Science': [89, 94, 82, 91, 87],
    'English': [78, 85, 88, 90, 92]
}
students_df = pd.DataFrame(data)
```

**Step 3: Descriptive Statistics**

Use the describe() method to generate a summary of descriptive statistics. This method provides the count, mean, standard deviation, minimum, and maximum values, as well as quartiles for each numerical column.

```python
summary = students_df.describe()
```

**Step 4: Custom Aggregations**

Pandas also allows for custom aggregations using the agg() method. For example, to calculate the mean and median of each subject:

```python
custom_summary = students_df.agg({
    'Math': ['mean', 'median'],
    'Science': ['mean', 'median'],
    'English': ['mean', 'median']
})
```

**Step 5: Grouped Summaries  **

For more complex analyses, data can be grouped using the groupby() method. Suppose we want to group students by their performance level and calculate the mean for each group:

```python
students_df['Performance'] = ['Good', 'Excellent', 'Average', 'Good', 'Excellent']

# Note: numeric_only=True is added to prevent a TypeError when trying to calculate the mean of non-numeric columns like 'Name'
grouped_summary = students_df.groupby('Performance').mean(numeric_only=True)
```

# Attributes and Methods of Pandas for Data Summary

Pandas provides a rich set of attributes and methods for data summary:   

* **mean():** Calculates the mean of DataFrame columns.   

* **median():** Computes the median of DataFrame columns.   

* **mode():** Returns the mode(s) of DataFrame columns.

* **std():** Computes the standard deviation of DataFrame columns.

* **var():** Calculates the variance of DataFrame columns.

* **sum():** Returns the sum of DataFrame columns.

* **min() and max():** Find the minimum and maximum values, respectively.

* **count():** Counts non-null entries in each column.

# Example: Analyzing Exam Scores

Let's revisit our example of Indian students' exam scores. Here's how we can use Pandas to summarize this data in a single, clean, and working script:

```python
import pandas as pd

# 1. Loading the Data
data = {
    'Name': ['Aarav', 'Isha', 'Rahul', 'Sneha', 'Ananya'],
    'Math': [85, 92, 78, 88, 95],
    'Science': [89, 94, 82, 91, 87],
    'English': [78, 85, 88, 90, 92]
}
students_df = pd.DataFrame(data)

# 2. Descriptive statistics
summary = students_df.describe()
print("--- Descriptive Statistics ---")
print(summary, "\n")

# 3. Custom aggregation
custom_summary = students_df.agg({
    'Math': ['mean', 'median'],
    'Science': ['mean', 'median'],
    'English': ['mean', 'median']
})
print("--- Custom Aggregation ---")
print(custom_summary, "\n")

# 4. Grouped summary
students_df['Performance'] = ['Good', 'Excellent', 'Average', 'Good', 'Excellent']
grouped_summary = students_df.groupby('Performance').mean(numeric_only=True)

print("--- Grouped Summary by Performance ---")
print(grouped_summary)
```

# Explaining the Code

In the code above, we first import Pandas and create a DataFrame with students' exam scores. We then use the *describe()* method to generate descriptive statistics, providing a quick overview of the dataset. Next, we perform custom aggregation to calculate the mean and median for each subject. Finally, we demonstrate how to group data by performance level and calculate the mean scores for each group.

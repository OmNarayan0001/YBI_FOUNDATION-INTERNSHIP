# **Basics of Linear Regression Model**

## **Introduction**

Linear Regression is a fundamental statistical and machine learning technique used to understand the relationship between variables (explanatory model) and to make predictions (predictive model). It models the relationship between one dependent variable (y) and one or more independent variables (X) by fitting a linear function to the observed data.

## **Equation of a Straight Line**

The equation of a straight line is a mathematical representation of a linear relationship between two variables. It is given by:

* **Y = mX + c**

  * **y:** Dependent variable (output or response).

  * **X:** Independent variable (input or predictor).

  * **m:** Slope of the line (rate of change of y with respect to x). Indicates the steepness and direction of the line.

    * Positive m: Line slopes upwards.

    * Negative m: Line slopes downwards.

  * c: Intercept (the value of y when X = 0). Represents the point where the line crosses the y-axis.

## **Linear Regression Equation**

* $Y = b_0 + b_1X_1 + b_2X_2 + ... + b_nX_n + error$

  1. Dependent Variable (y): The variable being predicted or explained.

  2. Independent Variable (X): The variable(s) used to predict the dependent variable.

  3. Intercept ($b_0$): The value of y when X is 0. It represents the point where the regression line crosses the Y-axis. May not always have a practical interpretation if X = 0 is outside the range of observed data.

  4. Slope ($b_1, b_2, \dots, b_n$): The change in y for a one-unit change in X. It indicates the steepness and direction of the relationship. A positive slope indicates a positive relationship, while a negative slope indicates a negative relationship.

  5. $\varepsilon$ (error term): This represents the difference between the actual value of y and the predicted value of y. It accounts for the variability in the data that is not explained by the linear relationship.

* **Simple Linear Regression (one independent variable)**

  * $Y = b_0 + b_1X_1 + error$

  * Represents the equation of best fit line that minimizes the error between the observed data points and the predicted values.

* **Multiple Linear Regression (with two independent variables)**

  * $Y = b_0 + b_1X_1 + b_2X_2 + error$

  * Represents the equation of best fit plane that minimizes the error between the observed data points and the predicted values.

* **Multiple Linear Regression (with more than two independent variables)**

  * $y = b_0 + b_1X_1 + b_2X_2 + \dots + b_nX_n + error$

  * Represents the equation of best fit hyper-plane that minimizes the error between the observed data points and the predicted values.

## **Least Squares Method**

The most common method for estimating the coefficients ($b_0$ and $b_1$). It minimizes the sum of the squared differences between observed values ($y_i$) and predicted values ($\hat{y}_i$):

* $\text{minimize} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$

## **Limitations of Linear Regression**

1. Assumes a linear relationship, which may not hold in real-world data.

2. Sensitive to outliers, which can distort the regression line.

3. Assumes that the predictors are uncorrelated (in simple linear regression, there's only one predictor, so this issue arises in multiple regression).

4. Limited to numerical independent variables or categorical variables with proper encoding.

## **Advantages**

1. Simplicity and Interpretability

2. Computational Efficiency

## **Disadvantages**

1. Sensitivity to Outliers

2. Assumption Dependencies

3. Limitation to Linear Relationships

## **Applications of Linear Regression**

1. Business Forecasting

2. Economics and Finance

3. Biological Sciences

4. Engineering Fields

# **Simple Linear Regression**

Simple linear regression is a statistical method used to model the relationship between a single independent variable (x) and a dependent variable (y). It assumes that there is a linear relationship between the independent and dependent variables.

### Formula:

The equation for simple linear regression can be represented as:

$y = \beta_0 + \beta_1x + \varepsilon$

### Assumptions:

  1. Linearity: There is a linear relationship between the independent and dependent variables.

  2. Independence: The observations are independent of each other.

  3. Homoscedasticity: The variance of the residuals is constant across all levels of the independent variable.

  4. Normality: The residuals are normally distributed.

  5. No Multicollinearity: There is no multicollinearity among the independent variables.

### **Advantages:**

* Provides insights into the relationship between variables.

* Simple and easy to interpret.

* Useful for prediction and forecasting.
### **Disadvantages:**
* Assumes a linear relationship which may not always be the case.

* Sensitive to outliers.

* Cannot capture complex relationships between variables.

### **Application:**

Simple linear regression is commonly used in various fields, including:

 * Economics: Predicting the impact of a change in one variable on another (e.g., predicting sales based on advertising expenditure).
 * Finance: Analyzing the relationship between intrest rates and stock prices.
 * Healthcare: Predicting patient outcomes based on treatements variables.

 ### **Sample Python Code (Simple Linear Regression):**

```python
# Importing libraries
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Generating sample data
np.random.seed(2529)
x = 2 * np.random.rand(100, 1)
y = 4 + 3 * x + np.random.randn(100, 1)

# Splitting the data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=2529)

# Creating a linear regression model
model = LinearRegression()

# Fitting the model
model.fit(x_train, y_train)

# Making predictions
y_pred = model.predict(x_test)

# Visualizing the results
plt.scatter(x_test, y_test, color='blue')
plt.plot(x_test, y_pred, color='red')
plt.title('Simple Linear Regression')
plt.xlabel('Independent Variable')
plt.ylabel('Dependent Variable')
plt.show()
```

# Multiple Linear Regression

Multiple linear regression (MLR) is a statistical technique that uses several explanatory variables to predict the outcome of a response variable. The goal of MLR is to model the linear relationship between the dependent (response) variable and more than one independent (explanatory) variable.

### Formula

The formula for a multiple linear regression model is:

$$y = \beta_0 + \beta_1x_1 + \beta_2x_2 + \dots + \beta_nx_n + \varepsilon$$

Where:

* $y$ is the dependent variable.

* $x_1, x_2, \dots, x_n$ are the independent variables

* $\beta_0$ is the intercept.

* $\beta_1, \beta_2, \dots, \beta_n$ are the coefficients of the independent variables.

* $\varepsilon$ is the error term, assumed to be normally distributed with mean zero.

### **How It Works**

Multiple linear regression works by estimating the coefficients ($\beta$) that best fit the linear relationship between the dependent variable and multiple independent variables. This fitting process typically involves minimizing the sum of the squares of the residuals (difference between observed and predicted values).

### **Implementation Steps**

1. Data Collection: Gather the data for the dependent variable and all independent variables.

2. Model Specification: Identify which variables will be used as the predictors and which variable as the response.

3. Assumption Checking: Check for linearity, multicollinearity, normality, homoscedasticity, and independence of residuals.

4. Fit the Model: Use statistical software to fit the model. Estimate the coefficients.

5. Model Evaluation: Evaluate the model using $R^2$, adjusted $R^2$, and p-values for the coefficients.

6. Model Refinement: Refine the model by possibly adding interaction terms, polynomial terms, or checking for potential influential points.

### **Assumptions**

* **Linearity:** The relationship between the dependent and independent variables is linear.

* *No multicollinearity:* Independent variables should not be too highly correlated with each other.

* **Homoscedasticity:** Constant variance of error terms.

* **Normality of Residuals:** Residuals should follow a normal distribution.

* **Independence:** Observations should be independent of each other.

### **Advantages**

* **Flexibility:** Can accommodate multiple influencing factors simultaneously.

* **Efficiency:** More efficient use of data as it explains more variance using multiple predictors.

* **Control for Confounders:** Ability to control for confounding variables.

### **Disadvantages**

* **Complexity:** More complex to understand and interpret compared to simple linear regression.

* **Risk of Overfitting:** Especially with many predictors or when data points are few.

* **Multicollinearity:** Can be a serious issue, making the model unstable and the interpretation of coefficients challenging.

### Applications

* **Economics:** Modeling economic growth by considering multiple factors like investment, consumption, and government spending.

* **Healthcare:** Predicting patient outcomes based on multiple characteristics like age, weight, and treatment type.

* **Real Estate:** Estimating property prices based on features like location, size, and number of rooms.

### **Sample Python Code (Statsmodels):**

Below is an example of how to implement multiple linear regression in Python using thr `statsmodles` library:

```python
import statsmodels.api as sm
import numpy as np

# Sample data
X = np.array([[1, 1], [1, 2], [2, 2], [2, 3]])  # multiple independent variables
Y = np.array([6, 8, 9, 11])  # dependent variable

# Adding a constant to the model (for the intercept)
X = sm.add_constant(X)

# Fitting the model
model = sm.OLS(Y, X).fit()

# Summary of the model
print(model.summary())
```

This code snippit demonstrates that basic steps for fitting a multiple linear regression model using Python. It includes data preparation, model fitting, and printout a summary that provides detailed statistics about the model's performance.

### **Sample Python Code (Scikit-Learn):**

Here's a sample Python code that demonstrates how to implement multiple linear regression using the sklearn library. This example uses a dataset with multiple independent variables to predict a dependent variable.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Sample dataset
# Let's assume a dataset with 3 features (X1, X2, X3) and one target variable (Y)
data = {
    'X1': [1, 2, 3, 4, 5],
    'X2': [5, 3, 6, 9, 8],
    'X3': [7, 4, 3, 9, 6],
    'Y': [14, 24, 18, 34, 17]
}
df = pd.DataFrame(data)

# Independent variables
X = df[['X1', 'X2', 'X3']]

# Dependent variable
y = df['Y']

# Splitting the dataset into the Training set and Test set
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2529)

# Creating a Linear Regression model
model = LinearRegression()

# Fitting the Linear Regression model
model.fit(X_train, y_train)

# Making predictions
y_pred = model.predict(X_test)

# Evaluating the model
mean_squared_error(y_test, y_pred)
r2_score(y_test, y_pred)
```

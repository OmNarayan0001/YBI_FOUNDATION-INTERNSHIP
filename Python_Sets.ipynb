# Python Sets
Set is a built-in Python data type used to store a collection of unique items.
* **Unordered collection:** Elements do not have a fixed position and cannot be accessed using indexes.
* **Automatic uniqueness:** Sets only store unique elements; duplicate values are instantly removed.
* **Mutability rules:** While the set itself is mutable (items can be added or removed), the elements inside it must be immutable and hashable (e.g., strings, numbers, tuples). You cannot store mutable data types like lists or dictionaries.
* **Mixed data types:** Sets are *Heterogeneous* and can store multiple different data types within the same collection.
* **High performance:** Backed by internal hashing, sets provide fast search, insertion, deletion, and membership testing, and are highly optimized for mathematical operations.

```python
# Demonstrating Unique, Immutable, and Heterogeneous properties
my_set = {1, 2, 2, 3, "Apple", (10, 20)}
print(my_set)
```

```python
# Elements inside a set MUST be immutable (specifically, hashable).
# You cannot insert lists, dictionaries, or other sets into a set.
try:
    invalid_set = {[1, 2], 3}  # Raises TypeError because list is unhashable
except TypeError as e:
    print("1B. Immutability Constraint Error:", e)

# Internal Working of Sets
# Internally, Python sets are implemented using a Hash Table.
print(f"1C. Internal Hash example for 'Python': {hash('Python')}")
```

# Type Casting
* **Type Casting** is the `set()` constructor is used to convert other iterables and data types (such as lists, tuples, or strings) into sets, this conversion technique is frequently used as a quick way to eliminate duplicate values from a list.

```python
# TYPE CASTING (CREATION)
my_list = [1, 2, 2, 3]
set_from_list = set(my_list)
print("\n2. Type Casting (List -> Set):", set_from_list)

set_from_string = set("banana")
print("2. Type Casting (String -> Set):", set_from_string)
```

# Frozen Sets
* **Immutability:** A `frozenset` is an immutable version of a Python set created using the `frozenset()` constructor. Once created, its elements cannot be added, removed, or changed.
* **Hashability:** Because frozen sets are immutable and hashable, they can be safely used as keys in dictionaries or as elements within another standard set.
* **Supported operations:** While they cannot be modified, frozen sets still support non-mutating mathematical operations like union (`|`), intersection (`&`), difference (`-`), and symmetric difference (`^`).

```python
# FROZEN SETS
fs = frozenset([1, 2, 3, 4])
print("\n3. Frozen Set:", fs)
try:
    fs.add(5)  # Will raise AttributeError
except AttributeError as e:
    print("3. Frozen Set Immutability:", e)
```

# Internal working of Set
* **Hash Table Implementation:** Python sets are built on hash tables (similar to dictionaries), where elements act as keys paired with dummy values.
* **Memory Indexing:** When adding an element, Python uses the built-in `hash()` function to calculate a hash value, which serves as the memory index.
* **O(1) Performance:** This architecture enables highly efficient, average O(1) time complexity for insertions, deletions, and lookups (membership testing).
* **Collision Handling:** If multiple elements generate the same hash index, Python resolves the collision by storing them in the same bucket.
* **Immutability Requirement:** Set elements must be immutable because altering an element would change its hash value and break the internal memory mapping.

# Methods and Mathematical Operators for Sets
Sets and frozensets support both built-in methods and standard mathematical operators for comparing and combining collections.

1. **Adding Elements:**
You can add a single element using .add() or multiple elements using .update().

2. **Union (|):**
Returns a new set containing all items from both sets, excluding duplicates.

3. **Intersection (&):**
Returns a new set containing only the items that exist in both sets.

4. **Difference (-):**
Returns a new set with items present in the first set but not in the second.

5. **Symmetric Difference (^):**
When you compare two sets, it throws away everything they share and keeps only the items that are unique to each individual set.

6. **Clear:**
Empties the set completely, leaving set().

**Containment Check:** Determines if a specific, individual item exists inside a set. Using the `in` or `not in` operators, it evaluates membership and returns `True` or `False`.

**Equivalence:** Compares two entire sets to check if they contain the exact same elements. Using the `==` or `!=` operators, it evaluates content equality; since sets are unordered, the sequence in which the elements were added does not matter.

**Subsets & Supersets:** Evaluates how much two sets overlap and whether one entirely encompasses the other.
1. **Subset (`<=`):** Checks if every element in the first set is also present in the second set.
2. **Superset (`>=`):** Checks if the first set completely contains every element of the second set.
3. **Proper Subset / Superset (`<`, `>`):** Performs the same containment checks, but strictly requires that the two sets are not exactly equal (the larger set must have at least one extra, unshared element).

```python
# Methods and mathematical Operators
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print("Initial Set A:", A)
print("Initial Set B:", B)

# 1. Adding elements (add / update)
A.add(99)  # Adds a single element
A.update([100, 101])  # Adds multiple elements
print("\n4.1. After Adding/Updating A:", A)


print("---------------------------------------------------------")
A = {1, 2, 3, 4}  # Reverting set A for remaining operations
print("Set A:", A)
# 2. Union (Combines elements from both sets)
print("4.2. Union (Method):", A.union(B))

# 3. Intersection (Common elements)
print("4.3. Intersection (Method):", A.intersection(B))

# 4. Difference (Elements in A but not in B)
print("4.4. Difference (Method):", A.difference(B))

# 5. Symmetric Difference Operator (^)
symmetric = A ^ B
print("4.5. Symmetric Difference (A ^ B):", symmetric)

# 6. Clear (Removes all elements, leaving an empty set)
temp_set = {1, 2, 3}
temp_set.clear()
print("4.6. Clear (Method):", temp_set)



# Defining variables for the remaining Operators
s1 = {1, 2, 3, 4}
s2 = {5, 6, 7}
s3 = {1, 2}
s4 = {1, 2, 3, 4}

# Containment check
print("\nContainment:")
print(f"3 in s1 -> {3 in s1}")  # True
print(f"9 not in s1 -> {9 not in s1}")  # True

# Equivalence
print("\nEquivalence:")
print(f"s1 == s4 -> {s1 == s4}")  # True
print(f"s1 != s2 -> {s1 != s2}")  # True

# Subsets & Supersets
print("\nSubsets & Supersets:")
print("Var of s1:", s1)
print("Var of s2:", s2)
print("Var of s3:", s3)
print("Var of s4:", s4)
print(f"s3 <= s1 -> {s3 <= s1}")  # True (s3 is a subset of s1)
print(f"s3 < s1 -> {s3 < s1}")  # True (s3 is a proper subset of s1)
print(f"s1 >= s3 -> {s1 >= s3}")  # True (s1 is a superset of s3)
print(f"s1 > s3 -> {s1 > s3}")  # True (s1 is a proper superset of s3)
```

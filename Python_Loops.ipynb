# Python Loops

```python
# 1. For Loop
# Used for iterating over a sequence (like a list, tuple, dictionary, set, or string).
print("--- For Loop Example ---")
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"I like {fruit}")

# Using range() with a for loop
print("\n--- For Loop with range() ---")
for i in range(2,6):
    print(f"i")

# 2. While Loop
# Executed as long as a condition remains True.
print("\n--- While Loop Example ---")
count = 1
while count <= 5:
    print(f"Count is {count}")
    count += 1

# 3. Nested Loops
# A loop inside another loop. The outer loop finishes one iteration for every full run of the inner loop.
print("\n--- Nested Loops Example ---")
adj = ["red", "ripe"]
for a in adj:
    for f in fruits:
        print(f"{a} {f}")

# 4. Loop Control Statements
# break: Terminates the loop prematurely when a condition is met
# continue: Skips the rest of the current iteration and goes to the next
print("\n--- Loop Control Statements Example ---")
for number in range(1, 10):
    if number == 2:
        print("Skipping 2 using continue")
        continue  # Skip the rest of this iteration
    if number == 5:
        print("Encountered 5, doing nothing using pass")
        pass  # Placeholder, execution continues
    if number == 7:
        print("Breaking the loop at 7")
        break  # Terminate the loop
    print(f"Number: {number}")
```

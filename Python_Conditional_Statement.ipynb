# Python Conditional Statement

```python
print("Python Conditional Statements\n")

# Conditional statements allow you to execute different blocks of code based on whether certain conditions are true or false.
# The primary keywords used are 'if', 'elif' (else if), and 'else'.

# 1. The 'if' statement
# Executes a block of code if a condition is true.
print("--- if statement ---")
x = 10
if x > 5:
    print(f"x ({x}) is greater than 5")

print("\n")

# 2. The 'if-else' statement
# Executes one block of code if the condition is true, and another if it's false.
print("-- if-else statement --")
y = 3
if y % 2 == 0:
    print(f"y ({y}) is an even number")
else:
    print(f"y ({y}) is an odd number")

z = 4
if z % 2 == 0:
    print(f"z ({z}) is an even number")
else:
    print(f"z ({z}) is an odd number")

print("\n")

# 3. The 'if-elif-else' statement
# Allows you to check multiple conditions sequentially.
# 'elif' checks another condition if the previous 'if' or 'elif' conditions were false.
print("-- if-elif-else statement --")
score = 75
if score >= 90:
    print(f"Score: {score} -> Grade A")
elif score >= 80:
    print(f"Score: {score} -> Grade B")
elif score >= 70:
    print(f"Score: {score} -> Grade C")
elif score >= 60:
    print(f"Score: {score} -> Grade D")
else:
    print(f"Score: {score} -> Grade F")

print("\n")

# Example with different score:
score = 92
if score >= 90:
    print(f"Score: {score} -> Grade A")
elif score >= 80:
    print(f"Score: {score} -> Grade B")
elif score >= 70:
    print(f"Score: {score} -> Grade C")
elif score >= 60:
    print(f"Score: {score} -> Grade D")
else:
    print(f"Score: {score} -> Grade F")

print("\n")

# 4. Nested 'if' statements
# An 'if' or 'else' block can contain other 'if' statements.
print("-- Nested if statements --")
age = 25
has_license = True

if age >= 18:
    print(f"Age: {age} -> Eligible to drive based on age.")
    if has_license:
        print("And has a valid license. Can drive!")
    else:
        print("But does not have a license. Cannot drive yet.")
else:
    print(f"Age: {age} -> Not old enough to drive.")

print("\n")

# Example with different values:
age = 16
has_license = False

if age >= 18:
    print(f"Age: {age} -> Eligible to drive based on age.")
    if has_license:
        print("And has a valid license. Can drive!")
    else:
        print("But does not have a license. Cannot drive yet.")
else:
    print(f"Age: {age} -> Not old enough to drive.")

print("\n")

# Important: Indentation is crucial in Python to define code blocks.
# Conditions can involve comparison operators (==, !=, <, >, <=, >=)
# and logical operators (and, or, not).

print("-- Conditional with Logical Operators --")
is_sunny = True
is_weekend = False

if is_sunny and is_weekend:
    print("It's a sunny weekend! Perfect for outdoors.")
elif is_sunny and not is_weekend:
    print("It's sunny, but it's a weekday. Time to work!")
elif not is_sunny and is_weekend:
    print("It's the weekend, but not sunny. Maybe relax indoors.")
else:
    print("It's a gloomy weekday. Stay productive!")
```

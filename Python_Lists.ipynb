# Python Lists
List is a built-in data structure used to store an ordered collection of items. They are dynamic, resizable and capable of storing multiple data types.

* List elements can be changed, updated, added, or removed after the list is created.

* Elements maintain the order in which they are inserted.

* Elements are accessed using their position, starting from index 0.

#Creating a List
Lists can be created in several ways, such as using square brackets [], the list() constructor or by repeating elements.

# **1. Square Brackets**
The most common way to create a list is by enclosing comma-separated elements in square brackets []. Lists are heterogeneous, meaning they can contain items of different data types (integers, strings, booleans, etc.).

```python
print("--- Square Brackets ---")
empty_list = []
fruits = ["apple", "banana", "cherry"]
mixed_list = [1, "hello", 3.14, True]
print("Empty:", empty_list)
print("Fruits:", fruits)
print("Mixed:", mixed_list)
print()
```

# **2. List() Constructor**
The list() built-in function creates a list by converting an existing iterable object, such as a string, tuple, or set.

```python
print("--- List() Constructor ---")
chars = list("Python")
from_tuple = list((1, 2, 3))
print("From String:", chars)
print("From Tuple:", from_tuple)
print()
```

# **3. Creating List with Repeated Elements**
You can use the multiplication operator * to quickly initialize a list with a repeated sequence of elements. This is useful for creating lists of a specific size filled with default values.

```python
print("--- Creating List with Repeated Elements ---")
zeros = [0] * 5
repeated_pattern = [1, 2] * 3
print("Zeros:", zeros)
print("Repeated Pattern:", repeated_pattern)
print()
```

#Internal Representation of Lists
Python list stores references to objects, not the actual values directly.

* The list keeps memory addresses of objects like integers, strings or booleans.

* Actual objects exist separately in memory.

* Modifying a mutable object inside a list changes the original object.

* Reassigning an immutable object creates a new object instead of changing the old one.

# **1. Accessing Elements**
Lists are ordered, and elements are accessed using zero-based indexing. You can use negative indexing to access elements from the end of the list (-1 is the last element). Slicing [start:stop:step] allows you to access a specific sub-range of the list.

```python
print("--- Accessing Elements ---")
numbers = [10, 20, 30, 40, 50]
print("First Element (index 0):", numbers[0])
print("Last Element (index -1):", numbers[-1])
print("Slice (indices 1 to 3):", numbers[1:4])
print()
```

# **2. Adding Elements**
Lists are mutable. You can add elements using:

* **append(item):** Adds a single item to the end of the list.

* **insert(index, item):** Inserts an item at a specific index.

* **extend(iterable):** Appends elements from another iterable to the end.


```python
print("--- Adding Elements ---")
languages = ["Python", "Java"]
languages.append("C++")
print("After append:", languages)
languages.insert(1, "JavaScript")
print("After insert at index 1:", languages)
languages.extend(["Rust", "Go"])
print("After extend:", languages)
print()
```

# **3. Updating Elements**
You can modify an existing element or a range of elements by assigning a new value to their specific index or slice.

```python
print("--- Updating Elements ---")
colors = ["red", "green", "blue"]
colors[1] = "yellow"
print("After updating index 1:", colors)
print()
```

# **4. Removing Elements**
Elements can be removed using various methods depending on whether you know the value or the index:

* **remove(value):** Removes the first occurrence of the specified value.

* **pop(index):** Removes and returns the element at the specified index (defaults to the last element if no index is provided).

* **del:** Deletes an element or a slice by index.

* **clear():** Empties the entire list.

```python
print("--- Removing Elements ---")
items = ["pen", "pencil", "eraser", "ruler", "marker"]
items.remove("pencil")
print("After removing 'pencil':", items)
popped_item = items.pop(1)
print(f"Popped item at index 1 ('{popped_item}'):", items)
del items[0]
print("After deleting index 0:", items)
items.clear()
print("After clearing all elements:", items)
print()
```

# **5. Iterating Over Lists**
You can loop through the elements of a list using a for loop. If you need to access both the index and the value during iteration, use the enumerate() function.

```python
print("--- Iterating Over Lists ---")
animals = ["cat", "dog", "bird"]
print("Simple iteration:")
for animal in animals:
    print(f" - {animal}")

print("\nIteration with indices:")
for index, animal in enumerate(animals):
    print(f" Index {index}: {animal}")
print()
```

# **6. Nested Lists**
A list can contain other lists as its elements, creating a multi-dimensional structure (like a matrix). You access items in nested lists by chaining square brackets [][].

```python
print("--- Nested Lists ---")
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print("Matrix:", matrix)
print("First row:", matrix[0])
print("Element at row 1, col 2 (value 6):", matrix[1][2])
```

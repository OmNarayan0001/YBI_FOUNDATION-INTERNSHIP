# **Basics of K-means Clustering**

K-means clustering is a popular unsupervised machine learning algorithm used to partition a dataset into distinct groups, or clusters. The goal of K-means is to divide the data points into K clusters, where each data point belongs to the cluster with the nearest mean. This technique is widely used in various domains, including marketing, finance, and biology, to discover patterns and insights from data.

K-means clustering is an iterative algorithm that aims to minimize the variance within each cluster while maximizing the variance between clusters. The algorithm assigns each data point to one of K clusters based on the similarity of the data points, which is often measured by the Euclidean distance from the cluster centroids.

# **Theory Behind K-means Clustering**

The K-means algorithm operates in the following steps:

1. **Initialization:** Choose K initial centroids randomly from the dataset. These centroids are the initial guesses for the centers of the clusters.

2. **Assignment:** Assign each data point to the nearest centroid based on the Euclidean distance. This forms K clusters.

3. **Update:** Recalculate the centroids of the newly formed clusters by taking the mean of all data points in each cluster.

4. **Repeat:** Repeat the assignment and update steps until the centroids no longer change significantly or a predefined number of iterations is reached.

# **Example Scenario**

Consider a scenario where an Indian retail company wants to segment its customers based on their purchasing behavior. The company has data on the annual spending of customers on various product categories. By applying K-means clustering, the company can identify distinct customer segments, such as budget shoppers, luxury buyers, and average spenders.

#### **Implementation Example**

Let's implement K-means clustering using Python and the popular machine learning library, scikit-learn. We'll use a dataset that contains information about customers' annual spending on various categories.

```python
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Sample data: Annual spending on different categories (in thousands)
data = {
    'Customer': ['Aarav', 'Ishita', 'Rohan', 'Sneha', 'Neha', 'Vikram', 'Ananya', 'Arjun'],
    'Spending_on_Electronics': [15, 20, 14, 25, 30, 18, 22, 24],
    'Spending_on_Clothing': [35, 40, 30, 45, 50, 38, 42, 44]
}
df = pd.DataFrame(data)

# Extracting features for clustering
X = df[['Spending_on_Electronics', 'Spending_on_Clothing']]

# Applying K-means clustering
kmeans = KMeans(n_clusters=3, random_state=0)
kmeans.fit(X)

# Adding cluster labels to the dataframe
df['Cluster'] = kmeans.labels_
print(df)

# Visualizing the clusters
plt.scatter(df['Spending_on_Electronics'], df['Spending_on_Clothing'], c=df['Cluster'], cmap='viridis')
plt.xlabel('Spending on Electronics (in thousands)')
plt.ylabel('Spending on Clothing (in thousands)')
plt.title('Customer Segmentation using K-means Clustering')
plt.show()
```

# **Explanation of the Code**

1. **Data Preparation:** We create a pandas DataFrame containing customer names and their annual spending on electronics and clothing.

2. **Feature Extraction:** We extract the relevant features for clustering, which are the spending on electronics and clothing.

3. **K-means Clustering:** We initialize the K-means algorithm with 3 clusters and fit it to the data.

4. **Cluster Assignment:** The algorithm assigns each customer to one of the 3 clusters based on their spending patterns.

5. **Visualization:** We visualize the clusters using a scatter plot, where different colors represent different clusters.

# **What is Machine Learning?**
Machine Learning (ML) is a subset of artificial intelligence that focuses on the development of algorithms and statistical models that enable computers to perform tasks without explicit instructions. It revolves around the idea that systems can learn from data, identify patterns, and make decisions with minimal human intervention.

# **Key concepts in machine learning:**

* **Data:** The foundation of machine learning. It involves collecting, processing, and analyzing vast amounts of data to train models.

* **Algorithms:** These are the mathematical instructions that guide the learning process. Common algorithms include linear regression, decision trees, and neural networks.

* **Models:** A model is the output of a machine learning algorithm after it has been trained on data. It is used to make predictions or decisions based on new data.

* **Training:** The process of feeding data into a machine learning algorithm to help it learn and improve its accuracy over time.

* **Testing and validation:** After training, models are tested using a separate dataset to evaluate their performance and ensure they generalize well to new, unseen data.

# **Types of machine learning:**

* **Supervised learning:** Involves training a model on a labeled dataset, meaning that each training example is paired with an output label. The model learns to predict the output from the input data. Examples include classification and regression tasks.

* **Unsupervised learning:** Involves training a model on data without labeled responses. The model tries to identify patterns and relationships in the data. Common techniques include clustering and dimensionality reduction.

* **Reinforcement learning:** Involves training models to make sequences of decisions by rewarding desired behaviors and punishing undesired ones. It is often used in robotics and game playing.

# **Applications of machine learning:**

* **Healthcare:** Predictive analytics for patient diagnosis, personalized treatment plans, and drug discovery.

* **Finance:** Fraud detection, algorithmic trading, and credit scoring.

* **Retail:** Personalized recommendations, inventory management, and customer sentiment analysis.

* **Transportation:** Autonomous vehicles, route optimization, and predictive maintenance.

* **Entertainment:** Content recommendation systems, such as those used by Netflix and Spotify.

### **Example 1: Supervised learning (Linear regression)**
This predicts an output (y) based on input data (X).

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Prepare data (X = inputs, y = outputs/labels)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10]) # y is basically 2 * X

# 2. Choose the algorithm and initialize the model
model = LinearRegression()

# 3. Train the model
model.fit(X, y)

# 4. Make a prediction with new, unseen data
new_data = np.array([[6]])
prediction = model.predict(new_data)

print(f"Prediction for input 6: {prediction[0]}")
# Output will be close to 12
```

### **Example 2: Unsupervised learning (K-Means clustering)**
This groups similar data points together without any predefined labels.

```python
import numpy as np
from sklearn.cluster import KMeans

# 1. Prepare data (only inputs, no labels)
X = np.array([[1, 2], [1, 4], [1, 0],
              [10, 2], [10, 4], [10, 0]])

# 2. Choose the algorithm (we want to find 2 clusters)
model = KMeans(n_clusters=2, random_state=0, n_init='auto')

# 3. Train the model to identify patterns
model.fit(X)

# 4. See which cluster each data point belongs to
print("Cluster assignments:", model.labels_)

# Predict the cluster for a new data point
new_point = np.array([[0, 0], [12, 3]])
print("Predictions for new points:", model.predict(new_point))
```

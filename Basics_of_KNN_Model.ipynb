# **Basics of KNN Model**

k-NN is a simple, non-parametric, lazy learning algorithm used for classification and regression. In both cases, the input consists of the k closest training examples in the feature space.

# **KNN for Classification**

* The output is a class membership. An object is classified by a plurality vote of its neighbors, with the object being assigned to the class most common among its k nearest neighbors.

# **KNN for Regression**

The output is the property value for the object. This value is the average of the values of its k nearest neighbors.

# **How KNN Works**

* **Choosing the right k:** The algorithm's performance heavily depends on the choice of the parameter k (number of nearest neighbors).

* **Distance Metrics:** Commonly used metrics are Euclidean, Manhattan, and Minkowski distances.

* **Majority Voting:** In classification, the label of the new point is determined by a majority vote of its nearest neighbors.

# **Steps to Implement k-NN**

* **Step 1:** Choose the number k of neighbors.

* **Step 2:** Calculate the distance from the query example to the labeled examples.

* **Step 3:** Sort the distance and determine nearest neighbors based on the k-th minimum distance.

* **Step 4:** Gather the categories of the nearest neighbors.

* **Step 5:** For classification, use simple majority of the category of nearest neighbors. For regression, calculate the mean or median.

* **Step 6:** Return the predicted class.

# **Choosing a Good Value for k**

* **Trade-off:** A smaller k can lead to noise having a higher influence on the result, and a larger k makes computation expensive and may include points from other classes.

* **Cross-validation:** One way to select an appropriate k is by using cross-validation. Typically, larger values of k reduce the effect of noise on the classification, but make boundaries between classes less distinct.

# **Advantages of k-NN**

* Simple to understand and implement.

* No need to build a model, tune several parameters, or make additional assumptions.

* The algorithm adapts as we collect new training data.

# **Disadvantages of k-NN**

* The algorithm gets significantly slower as the number of examples and/or predictors/independent variables increase.

* Sensitive to the local structure of the data.

* High memory requirement as it needs to store all of the training data.

* Performance depends on the correct setting of k and the distance metric.

# **Applications of k-NN**

* Banking system for predicting whether a person is fit for loan approval based on financial history.

* Politics, for classifying potential voters in two or more classes to predict who will vote for which candidate.

* Calculating credit ratings by comparing with persons of similar financial history.

# **Practical Tips**

* Always normalize or standardize data as k-NN is sensitive to the scaling of the data.

* Handle missing values to prevent bias.

* Consider weighted voting based on distance to reduce the influence of outliers.

# **Python Code Examples (using scikit-learn)**

Here are basic examples of how to implement KNN for both classification and regression using Python's scikit-learn library. As mentioned in the practical tips, standardizing the data is crucial for KNN.

### **1. KNN for Classification**


```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Sample data: Features (X) and Labels (y)
X = np.array([[1, 2], [1.5, 1.8], [5, 8], [8, 8], [1, 0.6], [9, 11]])
y = np.array([0, 0, 1, 1, 0, 1])

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Standardize the data (Important for KNN)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Initialize and train the KNN Classifier (using k=3)
knn_classifier = KNeighborsClassifier(n_neighbors=3)
knn_classifier.fit(X_train_scaled, y_train)

# Make predictions
y_pred = knn_classifier.predict(X_test_scaled)

print(f"Predictions: {y_pred}")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
```

### **2. KNN for Regression**

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error

# Sample data: Features (X) and Target Values (y)
X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([1.2, 1.9, 3.1, 4.0, 5.1, 5.8])

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Standardize the data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Initialize and train the KNN Regressor (using k=2)
knn_regressor = KNeighborsRegressor(n_neighbors=2)
knn_regressor.fit(X_train_scaled, y_train)

# Make predictions
y_pred = knn_regressor.predict(X_test_scaled)

print(f"Predictions: {y_pred}")
print(f"Mean Squared Error: {mean_squared_error(y_test, y_pred):.2f}")
```

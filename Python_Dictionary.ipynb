# Python Dictionary
A dictionary is a A collection of key-value pairs used to store data values like a map. Unlike sequences that hold single values, dictionaries associate unique, immutable keys (like strings or numbers) with specific values. They are mutable (changeable), maintain insertion order (as of Python 3.7), and do not allow duplicate keys.

* **Mutable:** You can change, add, or remove items after the dictionary is created.

* **Ordered:** As of Python 3.7, dictionaries maintain insertion order.

* **No Duplicates:** Dictionaries cannot have two items with the same key. If you assign a value to an existing key, it overwrites the old value.

* **Key Requirements:** Keys must be of an immutable data type (like strings, numbers, or tuples). Values can be of any data type.

# Creating a Dictionary
The process of initializing a dictionary. This is typically done by placing a comma-separated list of key-value pairs enclosed in curly braces {} or by using the built-in dict() constructor function.

```python
# 1. Using curly braces
student = {
    "name": "Om",
    "age": 21,
    "major": "Computer Application"
}

# 2. Using the dict() constructor
car = dict(brand="Ford", model="Mustang", year=1964)

# 3. Creating an empty dictionary
empty_dict = {}
```

# Accessing Elements
The method of retrieving a value from a dictionary by referencing its specific key. You can do this using square bracket notation (dictionary["key"]), which raises an error if the key doesn't exist, or by using the .get() method, which safely returns a default value if the key is not found.

```python
student = {"name": "Om", "age": 21}

# Method 1: Using square brackets
print(student["name"])
# Note: This will raise a KeyError if the key does not exist.

# Method 2: Using the get() method (Safer)
print(student.get("age"))
print(student.get("grade", "Not Found"))
```

# Adding and Modifying Elements
The process of inserting new data or updating existing data. Both actions use the same syntax: assigning a value to a specific key (e.g., user["role"] = "Admin"). If the key exists, the value is updated; if it doesn't, a new key-value pair is added. You can also use the .update() method to change multiple items at once.

```python
user = {"name": "krish", "role": "Admin"}

# Modifying an existing value
user["role"] = "SuperAdmin"

# Adding a new key-value pair
user["status"] = "Active"

# Using the update() method to add/modify multiple items at once
user.update({"role": "User", "last_login": "Today"})

print(user)
```

# Removing Elements
**Removing:** Deleting data from the dictionary using methods like *.pop()* (removes a specific key and returns its value), *.popitem()* (removes the last inserted pair), *the del keyword*, or *.clear()* to empty the entire dictionary.

```python
inventory = {"apples": 5, "bananas": 10, "oranges": 3, "grapes": 8}

# 1. pop(key): Removes the specified key and returns its value
removed_item = inventory.pop("bananas")
print(removed_item)

# 2. popitem(): Removes and returns the last inserted key-value pair as a tuple
last_item = inventory.popitem()
print(last_item)

# 3. del keyword: Removes the item with the specified key name
del inventory["apples"]

# 4. clear(): Empties the entire dictionary
inventory.clear()
print(inventory)
```

# Extracting Keys, Values, and Items
**Extracting:** Using built-in methods like *.keys()*, *.values()*, and *.items()* to return dynamic view objects of the dictionary's keys, its values, or its key-value pairs (as tuples), respectively.

```python
settings = {"theme": "Dark", "notifications": True, "volume": 80}

# Get all keys
print(settings.keys())

# Get all values
print(settings.values())

# Get all key-value pairs as tuples
print(settings.items())
```

# Iterating Through a Dictionary
Using a for loop to systematically go through the dictionary's contents. You can choose to iterate over just the keys (the default behavior), just the values, or both simultaneously using the .items() method.

```python
scores = {"Math": 90, "Science": 85, "English": 88}

# Loop through keys only (default behavior)
for subject in scores:
    print(subject)

# Loop through values only
for score in scores.values():
    print(score)

# Loop through both keys and values
for subject, score in scores.items():
    print(f"{subject}: {score}")
```

# Dictionary Comprehension
A concise, single-line syntactic construct used to generate a new dictionary from an iterable. It works similarly to list comprehensions, allowing you to apply expressions and conditional logic to build a dictionary dynamically.

```python
# Create a dictionary of squares for numbers 1 to 5
squares = {x: x**2 for x in range(1, 6)}
print(squares)

# Create a dictionary with conditional logic
even_squares = {x: x**2 for x in range(1, 6) if x % 2 == 0}
print(even_squares)
```

# Nested Dictionaries
A data structure where a dictionary contains one or more other dictionaries as its values. This is particularly useful for representing complex, hierarchical data structures, similar to JSON format.

```python
company = {
    "employee_1": {
        "name": "Radha",
        "role": "Developer"
    },
    "employee_2": {
        "name": "Krishana",
        "role": "Designer"
    }
}

# Accessing nested data
print(company["employee_1"]["name"])  # Output: Radha

# Modifying nested data
company["employee_2"]["role"] = "Lead Designer"
```
